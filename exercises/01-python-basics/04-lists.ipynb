## Exercise 1 - Indexing

Use:

`stress_levels = [3, 6, 8, 2, 9]`

Without using a loop:

- Print the first value.
- Print the third value.
- Print the last value using a negative index.

````python
stress_levels = [3, 6, 8, 2, 9]

print("First value:", stress_levels[0])
print("Third value:", stress_levels[2])
print("Last value:", stress_levels[-1])
````

## Exercise 2 - Modify an Element

Use:

`sleep_hours = [6, 7, 5, 8]`

Change the second value from `7` to `7.5`.

## Exercise 3 - `append()`

Use:

`heart_rates = [72, 88, 95]`

Add `105` to the end of the list.

````python
heart_rates = [72, 88, 95]

heart_rates.append(105)

print(heart_rates)
````

## Exercise 4 - `remove()`

Use:

`stress_levels = [2, 5, 8, 9, 3]`

Remove the value `8` using `remove()`.

Then print the list.

````python
stress_levels = [2, 5, 8, 9, 3]

stress_levels.remove(8)

print(stress_levels)
````

## Exercise 5 - Basic Statistics

Use:

`sleep_hours = [6, 7.5, 5, 8, 6.5]`

Calculate and print:

- Minimum
- Maximum
- Total
- Average

Use:

`min()`

`max()`

`sum()`

`len()`

````python
sleep_hours = [6, 7.5, 5, 8, 6.5]

minimum = min(sleep_hours)
maximum = max(sleep_hours)
total = sum(sleep_hours)
average = total / len(sleep_hours)

print("Minimum:", minimum)
print("Maximum:", maximum)
print("Total:", total)
print("Average:", average)
````

## Exercise 6 - Membership

Use:

`stress_levels = [2, 4, 6, 8, 10]`

Check whether the value `8` exists in the list.

If it exists, print:

`Stress level 8 exists`

````python
stress_levels = [2, 4, 6, 8, 10]

if 8 in stress_levels:
    print("Stress level 8 exists")
````

## Exercise 7 - Slicing

Use:

`values = [10, 20, 30, 40, 50, 60]`

Print:

- The first three elements
- The last three elements
- Elements from index `1` up to, but not including, index `4`

````python
values = [10, 20, 30, 40, 50, 60]

print("First three:", values[:3])
print("Last three:", values[-3:])
print("Index 1 to 3:", values[1:4])
````

## Exercise 8 - Filter High Stress Values

Use:

`stress_levels = [2, 8, 5, 9, 3, 7, 1, 10]`

Create an empty list:

`high_stress_levels = []`

Use a `for` loop to add every value greater than or equal to `7`.

The expected result is:

`[8, 9, 7, 10]`

````python
stress_levels = [2, 8, 5, 9, 3, 7, 1, 10]

high_stress_levels = []

for stress in stress_levels:
    if stress >= 7:
        high_stress_levels.append(stress)

print(high_stress_levels)
````

## Exercise 9 - Count Classes

Suppose:

`labels = [0, 1, 1, 0, 1, 0, 1, 1]`

Where:

- `0` = Normal
- `1` = Stress

Use `count()` to calculate:

- Number of normal cases
- Number of stress cases

````python
labels = [0, 1, 1, 0, 1, 0, 1, 1]

normal_cases = labels.count(0)
stress_cases = labels.count(1)

print("Number of normal cases:", normal_cases)
print("Number of stress cases:", stress_cases)
````
