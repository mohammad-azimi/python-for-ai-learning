## Exercise 1 - Positive, Negative, or Zero

Ask the user to enter a number.

Print:

- `Positive` if the number is greater than zero
- `Negative` if the number is less than zero
- `Zero` if the number is zero

````python
number = float(input("Enter a number: "))

if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")
````

## Exercise 2 - Age Classification

Ask the user to enter their age.

Print `Adult` if the age is at least 18.

Otherwise, print `Minor`.

````python
age = int(input("Enter your age: "))

if age >= 18:
    print("Adult")
else:
    print("Minor")
````

## Exercise 3 - Stress Level

Ask the user to enter a stress score from 0 to 10.

Classification:

`0` to less than `4` → Low stress

`4` to less than `7` → Moderate stress

`7` to `10` → High stress

````python
stress_score = float(input("Enter your stress score: "))

if stress_score >= 7:
    print("High stress")
elif stress_score >= 4:
    print("Moderate stress")
else:
    print("Low stress")
````

## Exercise 4 - Sleep Duration

Ask the user to enter the number of hours they slept.

Print:

- `Not enough sleep` if sleep is less than 6 hours
- `Normal sleep` if sleep is between 6 and 8 hours
- `Long sleep` if sleep is more than 8 hours

````python
sleep_hours = float(input("Enter your sleep hours: "))

if sleep_hours < 6:
    print("Not enough sleep")
elif sleep_hours <= 8:
    print("Normal sleep")
else:
    print("Long sleep")
````

## Exercise 5 - Using `and`

Ask the user for:

- Stress score
- Sleep hours

If the stress score is at least `7` AND sleep duration is less than `6`, print `High risk`.

Otherwise, print `Normal risk`.

````python
stress_score = float(input("Enter your stress score: "))
sleep_hours = float(input("Enter your sleep hours: "))

if stress_score >= 7 and sleep_hours < 6:
    print("High risk")
else:
    print("Normal risk")
````

## Exercise 6 - Using `or`

Ask the user for:

- Heart rate
- Stress score

If the heart rate is greater than `100` OR the stress score is at least `8`, print `Attention needed`.

Otherwise, print `Condition is normal`.

````python
heart_rate = float(input("Enter your heart rate: "))
stress_score = float(input("Enter your stress score: "))

if heart_rate > 100 or stress_score >= 8:
    print("Attention needed")
else:
    print("Condition is normal")
````

## Exercise 7 - Simple Risk Assessment System

Create a small condition assessment program.

Inputs:

- `heart_rate`
- `sleep_hours`
- `stress_score`

Rules:

If the stress score is at least `8` AND sleep duration is less than `6`:

`High risk`

Otherwise, if the stress score is at least `5` OR the heart rate is greater than `95`:

`Moderate risk`

Otherwise:

`Low risk`

````python
heart_rate = float(input("Enter your heart rate: "))
sleep_hours = float(input("Enter your sleep hours: "))
stress_score = float(input("Enter your stress score: "))

if stress_score >= 8 and sleep_hours < 6:
    print("High risk")
elif stress_score >= 5 or heart_rate > 95:
    print("Moderate risk")
else:
    print("Low risk")
````
