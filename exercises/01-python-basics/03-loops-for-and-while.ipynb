## Exercise 1 - Print Numbers

Use a `for` loop to print the numbers from `1` to `10`.

````python
for number in range(1, 11):
    print(number)
````

## Exercise 2 - Even Numbers

Print the even numbers from `2` to `20`.

````python
for number in range(2, 21, 2):
    print(number)
````

## Exercise 3 - Stress Classification

Use the following stress values:

`[2, 8, 5, 9, 3, 7]`

For each value:

- Print `High stress` if the value is at least `7`.
- Otherwise, print `Normal stress`.

````python
stress_levels = [2, 8, 5, 9, 3, 7]

for stress in stress_levels:
    if stress >= 7:
        print("High stress")
    else:
        print("Normal stress")
````

## Exercise 4 - Count High Stress Cases

Use:

`[2, 8, 5, 9, 3, 7]`

Count how many values are greater than or equal to `7`.

````python
stress_levels = [2, 8, 5, 9, 3, 7]

high_stress_count = 0

for stress in stress_levels:
    if stress >= 7:
        high_stress_count += 1

print("Number of high stress cases:", high_stress_count)
````

## Exercise 5 - Total and Average Sleep

Use:

`[6, 7.5, 5, 8, 6.5]`

Calculate the total using a loop.

Do not use `sum()`.

Then calculate the average.

````python
sleep_hours = [6, 7.5, 5, 8, 6.5]

total = 0

for sleep in sleep_hours:
    total += sleep

average = total / len(sleep_hours)

print("Total sleep:", total)
print("Average sleep:", average)
````

## Exercise 6 - `while` Loop

Create:

`number = 1`

Use a `while` loop to print the numbers from `1` to `5`.

````python
number = 1

while number <= 5:
    print(number)
    number += 1
````

## Exercise 7 - Heart Rate Data

Use:

`[72, 105, 88, 110, 95, 68]`

For each heart rate:

- Print `High heart rate` if the value is greater than `100`.
- Otherwise, print `Normal heart rate`.

At the same time, count the number of high heart rate cases.

````python
heart_rates = [72, 105, 88, 110, 95, 68]

high_heart_rate_count = 0

for heart_rate in heart_rates:
    if heart_rate > 100:
        print("High heart rate")
        high_heart_rate_count += 1
    else:
        print("Normal heart rate")

print(
    "Number of high heart rate cases:",
    high_heart_rate_count
)
````
