## Exercise 2 - Arithmetic Operations

Create two variables:

- `a = 15`
- `b = 4`

Calculate and print:

- Addition
- Subtraction
- Multiplication
- Division
- Exponentiation

```python
a = 15
b = 4

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Exponentiation:", a ** b)
```

## Exercise 3 - Average Sleep

Suppose the sleep duration for three days is:

- Day 1: `6.5` hours
- Day 2: `7` hours
- Day 3: `5.5` hours

Calculate the average sleep duration and store it in a variable called `average_sleep`.

```python
sleep_day1 = 6.5
sleep_day2 = 7
sleep_day3 = 5.5

average_sleep = (sleep_day1 + sleep_day2 + sleep_day3) / 3

print("Average sleep:", average_sleep)
```

## Exercise 4 - User Input

Ask the user:

**How many hours did you sleep last night?**

Store the value in a variable called `sleep_hours`.

Because sleep duration can contain decimal values such as `6.5`, convert the input to `float`.

```python
sleep_hours = float(
    input("How many hours did you sleep last night? ")
)

print("You slept", sleep_hours, "hours.")
```

## Exercise 5 - Simple AI-Related Data

Suppose we have the following measurements for one person:

- Heart rate: `90`
- Sleep hours: `5`
- Stress score: `8`

Calculate the average of these three values.

Store the result in a variable called `average_value`.

Then use `type()` to check the data type of the result.

```python
heart_rate = 90
sleep_hours = 5
stress_score = 8

average_value = (
    heart_rate + sleep_hours + stress_score
) / 3

print("Average value:", average_value)
print("Data type:", type(average_value))
```
